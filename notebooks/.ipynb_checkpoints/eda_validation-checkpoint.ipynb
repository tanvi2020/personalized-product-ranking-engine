```python
import os
import sys
import pandas as pd
import numpy as np
```

```python
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.append(project_root)

from src.ranking.ranker import ranking
```

```python
def load_products():
    path = os.path.abspath(
        os.path.join(os.getcwd(), "..", "data", "raw", "products.csv")
    )
    return pd.read_csv(path)
```

```python
def load_users():
    path = os.path.abspath(
        os.path.join(os.getcwd(), "..", "data", "raw", "users.csv")
    )
    return pd.read_csv(path)
```

```python
# Load data once
products_df = load_products()
users_df = load_users()
```

```python
# Pick one sample user from each persona
budget_user = users_df[users_df["persona_type"] == "Budget"].iloc[0]
quality_user = users_df[users_df["persona_type"] == "Quality"].iloc[0]
```

```python
# Run ranking once and reuse these results later
budget_ranked = ranking(products_df, budget_user)
quality_ranked = ranking(products_df, quality_user)
```

```python
def check_price_distribution():
    print("===== PRICE DISTRIBUTION =====")
    print(f"Under 500:  {(products_df['price'] <= 500).mean() * 100:.2f}%")
    print(f"Under 1000: {(products_df['price'] <= 1000).mean() * 100:.2f}%")
    print(f"Under 2000: {(products_df['price'] <= 2000).mean() * 100:.2f}%")
    print()
    print("The price distribution shows a reasonable spread across low, medium and higher price ranges"
          "roughly thirty percent of products cost ₹500 or less, about fifty‑six percent are ₹1 000 or less "
          "and about seventy‑four percent are under ₹2 000. This suggests a balanced mix of affordable"
          "and mid‑range items, which supports diverse recommendations.")
```

```python
def check_rating_distribution():
    print("===== RATING DISTRIBUTION =====")
    print(products_df["rating"].value_counts(dropna=False).sort_index())
    print()

    print("Observation:")
    print(
        "Ratings are mostly skewed toward higher values, which is typical in user‑generated product data."
        "This skew is not problematic on its own because there is still a variety of ratings across the scale,"
        "and it does not appear to dominate the logic unduly.."
    )
    print()
```

```python
def check_purchases_vs_rating():
    print("===== AVERAGE PURCHASES PER RATING =====")
    print(products_df.groupby("rating")["num_purchases"].mean())
    print()

    print("Observation:")
    print(
        "Examining average purchases per rating reveals that products with higher ratings"
        "tend to have more purchases, but the pattern is moderate rather than extreme."
        "The dataset therefore reflects the intuitive notion that better products sell" 
        "more while leaving room for other factors such as price and brand to influence popularity."
    )
    print()
```

```python
def check_rating_purchase_correlation():
    print("===== CORRELATION: PURCHASES VS RATING =====")
    corr = products_df[["num_purchases", "rating"]].corr()
    print(corr)
    print()

    print("Observation:")
    print(
        "The correlation of 0.424 between purchases and rating indicates a moderate positive relationship."
        "This means that popularity and rating are related, but popularity is not being driven solely by rating,"
        "which is a healthier scenario for recommendation logic."
    )
    print()
```

```python
def check_missing_ratings():
    print("===== MISSING RATING CHECK =====")
    missing_percent = products_df["rating"].isna().mean() * 100
    print(f"Missing ratings (%): {missing_percent:.2f}")
    print()

    print("Observation:")
    print(
        "Approximately twenty‑one percent of products have missing ratings,"
        "a realistic proportion in a marketplace where newer or niche items may not yet have been reviewed."
        "This helps the ranking algorithm test how it handles items with incomplete information."
    )
    print()
```

```python
def check_review_count_distribution():
    print("===== REVIEW COUNT DISTRIBUTION =====")
    print(products_df["review_count"].describe())
    print(f"Mean review count: {products_df['review_count'].mean():.2f}")
    print()

    print("Observation:")
    print(
        "The review count distribution spans a wide range, from products with no reviews to products with over a hundred."
        "The mean is around forty‑one, but the median is thirty‑seven and the lower quartile is six, indicating"
        "that many products have only a handful of reviews. This spread of review counts reflects varied user engagement"
        "levels and supports nuanced ranking."
    )
    print()
```

```python
def check_rating_quality_correlation():
    print("===== CORRELATION: RATING VS INTRINSIC QUALITY =====")
    print(products_df[["rating", "intrinsic_quality"]].corr())
    print()

    print("Observation:")
    print(
        
    )
    print("Rating and intrinsic quality are almost perfectly correlated, suggesting that the ratings reflect"
          "underlying product quality very closely. This alignment means that better‑quality products generally have"
          "higher ratings, which is the intended relationship.")
```

```python
def check_top_10_products():
    columns_to_show = [
        "price",
        "price_score",
        "purchase_score",
        "trusted_rating_score",
        "trusted_adjusted_purchase_score",
        "final_score",
        "intrinsic_quality",
    ]

    print("========= TOP 10 PRODUCTS FOR BUDGET USER =========")
    print(budget_ranked[columns_to_show].head(10))
    print()

    print("========= TOP 10 PRODUCTS FOR QUALITY USER =========")
    print(quality_ranked[columns_to_show].head(10))
    print()

    print("Observation:")
    print(
        "The top‑ten products for the Budget user include items with high price scores and respectable final scores,"
        "reflecting a balance of affordability and demand. The top‑ten products for the Quality user have higher intrinsic"
        "quality scores and trusted rating scores, indicating that quality signals are being prioritised. "
        "The average intrinsic quality of both users’ top lists is significantly higher than the overall product average,"
        "showing that the ranking function is selecting stronger products for each persona."
        "A small number of items for the Quality user have lower quality or missing trust values;"
        "these are edge cases and could be further tuned, but overall the recommendations align well with expectations."
    )
    print()

    suspicious_quality_products = quality_ranked[
        (quality_ranked["intrinsic_quality"] < 0.6)
        | (quality_ranked["trusted_rating_score"].isna())
    ]

    print("Suspicious Quality-user products:")
    print(suspicious_quality_products.head(10))
    print()

    budget_avg_quality = budget_ranked["intrinsic_quality"].head(10).mean()
    quality_avg_quality = quality_ranked["intrinsic_quality"].head(10).mean()
    overall_avg_quality = products_df["intrinsic_quality"].mean()

    print("Average intrinsic quality of Budget top 10:", budget_avg_quality)
    print("Average intrinsic quality of Quality top 10:", quality_avg_quality)
    print("Overall average intrinsic quality:", overall_avg_quality)
    print()

```

```python
def dcg(scores):
    scores = np.array(scores)
    positions = np.arange(1, len(scores) + 1)
    return np.sum(scores / np.log2(positions + 1))


def idcg(scores, k):
    ideal_scores = sorted(scores, reverse=True)[:k]
    return dcg(ideal_scores)


def ndcg(scores, k):
    actual = dcg(scores[:k])
    ideal = idcg(scores, k)

    if ideal == 0:
        return 0

    return actual / ideal


def check_ndcg():
    budget_scores = budget_ranked["intrinsic_quality"].values
    quality_scores = quality_ranked["intrinsic_quality"].values

    print("===== NDCG SCORES =====")
    print("Budget NDCG@10:", ndcg(budget_scores, 10))
    print("Quality NDCG@10:", ndcg(quality_scores, 10))
    print()
```

```python
def final_summary():
    print("===== FINAL SUMMARY =====")
    print(
        "Finally, the NDCG scores of 0.85 for the Budget user and 0.88 for the Quality user demonstrate that the ranking"
        "function is producing high‑quality recommendations. Overall, the dataset and ranking logic now behave consistently:"
        "price, rating, purchases, reviews and intrinsic quality show realistic relationships,"
        "and the system is producing plausible top recommendations for different user profiles"
    )
    print()
```

```python
def run_all_checks():
    check_price_distribution()
    check_rating_distribution()
    check_purchases_vs_rating()
    check_rating_purchase_correlation()
    check_missing_ratings()
    check_review_count_distribution()
    check_rating_quality_correlation()
    check_top_10_products()
    check_ndcg()
    final_summary()


run_all_checks()
```
