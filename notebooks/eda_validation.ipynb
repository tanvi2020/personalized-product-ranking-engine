```python
import os
import pandas as pd
```

```python

def load_products():
    """
    Load the synthetic product dataset from the project data folder.
    Assumes this script is run from the notebooks/ or EDA/ directory.
    """
    base_dir = os.getcwd()
    file_path = os.path.abspath(
        os.path.join(base_dir, "..", "data", "raw", "products.csv")
    )
    return pd.read_csv(file_path)
```

```python
def check_price_distribution(df: pd.DataFrame) -> None:
    print("===== PRICE DISTRIBUTION =====")
    print(f"Under 500:  {(df['price'] <= 500).mean() * 100:.2f}%")
    print(f"Under 1000: {(df['price'] <= 1000).mean() * 100:.2f}%")
    print(f"Under 2000: {(df['price'] <= 2000).mean() * 100:.2f}%")
    print()
```

```python
def check_rating_distribution(df: pd.DataFrame) -> None:
    print("===== RATING DISTRIBUTION =====")
    rating_counts = df["rating"].value_counts(dropna=False).sort_index()
    print(rating_counts)
    print()

    print("Observation:")
    print(
        "Ratings are skewed toward 4-5, which is acceptable for marketplace data. "
        "The rating distribution itself is not the main issue."
    )
    print()
```

```python
def check_purchases_vs_rating(df: pd.DataFrame) -> None:
    print("===== AVERAGE PURCHASES PER RATING =====")
    avg_purchases = df.groupby("rating")["num_purchases"].mean()
    print(avg_purchases)
    print()

    print("Observation:")
    print(
        "Products with rating >= 4 have disproportionately high average purchases. "
        "This suggests that rating and popularity are too tightly coupled."
    )
    print(
        "In real marketplaces, popularity is influenced by multiple factors such as "
        "brand, exposure, age, and price, not just rating."
    )
    print()
```

```python
def check_rating_purchase_correlation(df: pd.DataFrame) -> None:
    print("===== CORRELATION: PURCHASES VS RATING =====")
    correlation = df[["num_purchases", "rating"]].corr()
    print(correlation)
    print()

    corr_value = correlation.loc["num_purchases", "rating"]
    print("Observation:")
    print(
        f"Correlation between purchases and rating is {corr_value} "
        "This relationship is stronger than expected and should be weakened."
    )
    print(
        "If rating and purchases carry nearly the same signal, the ranking system may "
        "double-count product quality and popularity."
    )
    print()
```

```python
def check_missing_ratings(df: pd.DataFrame) -> None:
    print("===== MISSING RATING CHECK =====")
    missing_rating_percentage = df["rating"].isna().mean() * 100
    print(f"Missing ratings (%): {missing_rating_percentage:.2f}")
    print()

    print("Observation:")
    print(
        "A low missing-rating percentage suggests that the rating visibility threshold "
        "is too lenient."
    )
    print(
        "Real marketplaces usually contain more unrated or weakly reviewed products, "
        "especially in the low-demand segment."
    )
    print()
```

```python

def check_review_count_distribution(df: pd.DataFrame) -> None:
    print("===== REVIEW COUNT DISTRIBUTION =====")
    review_stats = df["review_count"].describe()
    print(review_stats)
    print(f"Mean review count: {df['review_count'].mean():.2f}")
    print()

    print("Observation:")
    print(
        "The review-count distribution should include a stronger lower tail. "
        "Too many products currently fall into medium-review ranges."
    )
    print(
        "This indicates that the synthetic marketplace still lacks enough weak, "
        "low-engagement products."
    )
    print()
```

```python
def check_rating_quality_correlation(df: pd.DataFrame) -> None:
    print("===== CORRELATION: RATING VS INTRINSIC QUALITY =====")
    correlation = df[["rating", "intrinsic_quality"]].corr()
    print(correlation)
    print()

    print("Observation:")
    print(
        "Rating should remain positively correlated with intrinsic quality, since "
        "rating is intended to reflect underlying product quality."
    )
    print(
        "The goal is not to remove this relationship, but to prevent rating from "
        "becoming unrealistically deterministic."
    )
    print()
```

```python
def final_summary() -> None:
    print("===== FINAL SUMMARY =====")
    print(
        "- Rating distribution is acceptable, but rating and purchases are too strongly linked.\n"
        "- Missing ratings are a useful realism check and should be higher for low-demand products.\n"
        "- Review counts still need a stronger low-engagement tail.\n"
        "- The dataset is improving, but realism correction is still in progress.\n"
        "- Current priority: create weaker low-demand products rather than just increasing their count."
    )
```

```python

def main():
    df = load_products()

    check_price_distribution(df)
    check_rating_distribution(df)
    check_purchases_vs_rating(df)
    check_rating_purchase_correlation(df)
    check_missing_ratings(df)
    check_review_count_distribution(df)
    check_rating_quality_correlation(df)
    final_summary()


if __name__ == "__main__":
    main()
```
